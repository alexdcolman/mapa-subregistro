# Análisis geoespacial del no registro de ingresos en aglomerados urbanos de la EPH

## 1. Introducción

El índice de pobreza del segundo semestre de 2024 publicado en marzo de 2025 por el Instituto Nacional de Estadística y Censos de la República Argentina (INDEC) mostró una caída significativa con relación al mismo período del año anterior, en un contexto de recesión económica, contracción del consumo y deterioro sostenido de los ingresos reales. Esta aparente contradicción reavivó el debate en torno a la calidad de las estadísticas producidas a partir de la Encuesta Permanente de Hogares (EPH), en particular respecto del modo en que los ingresos son relevados, procesados y declarados en esa fuente.

Entre los distintos factores que pueden afectar la confiabilidad de las mediciones basadas en las variables de ingresos de la EPH, uno de los aspectos es su no declaración. Este fenómeno –frecuentemente eclipsado por la discusión sobre el subreporte (Schteingart, 2025; Tornarolli, mimeo)– se expresa en el registro de hogares con valor “0” en la variable “Ingreso Total Familiar” (ITF), aun cuando otros indicadores sugieren que podrían contar con recursos económicos. Según los datos del segundo trimestre de 2024, un 23% de los hogares relevados reporta un ITF igual a cero, una cifra significativa difícil de conciliar con la hipótesis de ausencia total de ingresos reales.

El análisis exploratorio de estos casos sugiere rechazar una lectura literal de los valores nulos. Los hogares con ITF = 0 tienden a compartir características con aquellos que sí declaran ingresos e incluso exhiben mayor frecuencia de atributos vinculados a niveles de vida medios o altos: mayor número de habitaciones de uso exclusivo, mayor proporción de viviendas en propiedad y presencia de ingresos laborales y no laborales (por trabajo, inversiones o alquileres, entre otros). De hecho, algunos estudios plantean la asociación de la no respuesta a las preguntas de ingresos (en particular, en la EPH en el Gran Buenos Aires entre 1990 y 2010) con ciertos perfiles: niveles altos de instrucción, patrones o empleadores, cuentapropistas, calificación profesional, tareas de comercialización, fuentes de ingresos laborales no asalariados y mixtos, entre otros. “En función de esto, puede presumirse que parte de los no declarantes, mayoritariamente, poseen niveles de ingresos altos y se ubican en los niveles elevados de la estatificación social lo cual sesgaría los estudios de atributos sociales basados en estratificaciones basadas en los ingresos. Además, otra parte importante de los no declarantes se encontraría entre los ocupados en subempleos indigentes sin ingresos fijos” (Donza, 2011: 13). El cruce entre las bases “hogares” y “personas” de la EPH refuerza esta hipótesis, al mostrar que muchos de los hogares con ITF = 0 sí informan ocupaciones o fuentes de ingresos individuales. Estos elementos sugieren que, lejos de reflejar situaciones de pobreza extrema, una proporción considerable de estos casos podría responder a omisiones en la declaración, problemas de captación o errores de registro, lo que configura un patrón de no respuesta sistemática que afecta la calidad de los datos.

Por otra parte, la distribución territorial de estos registros no es homogénea. En regiones como el Gran Buenos Aires (codificada “1” en la siguiente tabla), la proporción de hogares con ITF = 0 asciende al 37%, muy por encima del promedio nacional:

*Nota: frecuencias absolutas y proporciones de distribución de ITF = 0 en la EPH del segundo trimestre de 2024.*

Esta concentración sugiere la necesidad de indagar en factores estructurales y/o coyunturales como la informalidad laboral, las políticas fiscales o la confianza institucional, así también en aspectos metodológicos del relevamiento que podrían influir en la omisión. Cualquiera sea la causa, el resultado es una distorsión que compromete la estimación precisa de los niveles de ingresos, y, por esto, los indicadores sociales construidos a partir de ellos.

Este trabajo analiza geoespacialmente la distribución del no registro de ingresos en los aglomerados urbanos incluidos en la EPH. Se identifican patrones regionales y posibles zonas críticas mediante el uso de capas de referencia territoriales (provincias, departamentos, municipios), y, en algunos casos, la superposición de los datos del Registro Nacional de Barrios Populares (ReNaBaP) como indicador de vulnerabilidad estructural. A partir de herramientas estadísticas y computacionales, se busca aportar a una comprensión más precisa de las limitaciones territoriales de la EPH como fuente para la medición de ingresos y pobreza.

### 1.1. Problema e hipótesis

Los registros de hogares con ingreso total familiar igual a cero (ITF = 0) en la Encuesta Permanente de Hogares (EPH) del INDEC alcanzan niveles significativos que resultan difíciles de conciliar con la hipótesis de carencia absoluta de ingresos. La elevada proporción de estos casos, así como su distribución territorial desigual, sugiere la existencia de un patrón sistemático de no respuesta u omisión en la declaración de ingresos.

**Problema:** El no registro de ingresos en la EPH constituye un límite metodológico que afecta la confiabilidad de los indicadores construidos mediante esa fuente. Este fenómeno no parece distribuirse de forma aleatoria, sino que se concentra en ciertos aglomerados urbanos y regiones estadísticas, lo que plantea la necesidad de indagar en los factores territoriales y sociodemográficos que podrían estar incidiendo en su aparición. Comprender esta distribución desigual es clave para identificar zonas críticas, evaluar la representatividad de los datos y mejorar la interpretación de los niveles de ingreso y pobreza, sobre todo en contextos de alta informalidad y desconfianza institucional.

**Hipótesis:** La omisión de ingresos en la EPH no se distribuye aleatoriamente, sino que presenta patrones territoriales asociados a factores estructurales como la informalidad laboral, la desigualdad habitacional y la vulnerabilidad urbana. Se espera encontrar mayores proporciones de ITF = 0 en regiones con alta precariedad (Rojas Symmes, 2019) estructural, aunque también en zonas con perfiles socioeconómicos medios-altos, donde el no registro podría responder a lógicas de ocultamiento estratégico o desinterés.

### 3.2. Análisis Exploratorio de Datos

El Análisis Exploratorio de Datos (AED), dijimos, comprende una serie de etapas necesarias para comprender y preparar la información, orientadas a caracterizar el fenómeno del no registro de ingresos en la Encuesta Permanente de Hogares (EPH). Los pasos principales son los siguientes:

**1. Limpieza y revisión preliminar de variables clave**

Como parte fundamental del preprocesamiento, se procedió a la inspección y limpieza de las variables clave involucradas en el análisis, tanto en la base de hogares como en la base individual. Las variables seleccionadas para esta etapa fueron:

- En la base de hogares: AGLOMERADO, REGION (identificadores geográficos categóricos) e ITF (Ingreso Total Familiar, variable numérica).

- En la base individual: CH03 (relación de parentesco, categórica), CH06 (edad, numérica), CAT_OCUP (condición de actividad, categórica) y P47T (monto de ingreso total individual, variable numérica monetaria).

Para garantizar la calidad y coherencia de los datos, se realizaron análisis exploratorios preliminares. Para las variables categóricas, se calcularon las frecuencias absolutas de cada categoría, incluyendo valores faltantes o códigos especiales, con el objetivo de identificar posibles inconsistencias, errores o categorías poco representativas.

Para las variables numéricas, se generaron estadísticas descriptivas (como media, mediana, desviación estándar, mínimos y máximos) y se cuantificó la cantidad de valores nulos o potencialmente inválidos (por ejemplo, valores negativos o códigos utilizados para indicar datos faltantes).

Este proceso permite definir criterios objetivos para la limpieza y filtrado de datos, tales como la recodificación o eliminación de valores atípicos, la imputación o exclusión de casos con datos faltantes, y la normalización de categorías para facilitar el análisis posterior.
```python
# Variables a analizar

# En df_hogares

variables_hogares = ['AGLOMERADO', 'REGION', 'ITF']  # ITF es numérica pero puede tener valores especiales

# En df_personas

variables_personas_categoricas = ['CH03', 'CAT_OCUP']
variables_personas_numericas = ['P47T', 'CH06']

# Función para frecuencias de categóricas

def freq_categoricas(df, vars_cat):
    for var in vars_cat:
        print(f"Frecuencia para variable categórica '{var}':")
        print(df[var].value_counts(dropna=False).sort_index())
        print("-" * 40)

# Función para resumen de numéricas

def resumen_numericas(df, vars_num):
    for var in vars_num:
        print(f"Resumen para variable numérica '{var}':")
        print(df[var].describe())
        print(f"Cantidad de valores nulos en {var}: {df[var].isna().sum()}")
        if (df[var] < 0).any():
            print(f"Cantidad de valores negativos en {var}: {(df[var] < 0).sum()}")
        print("-" * 40)

# Ejecutar análisis

print("=== Análisis de df_hogares ===")
freq_categoricas(df_hogares, variables_hogares)

print("=== Análisis de df_personas ===")
freq_categoricas(df_personas, variables_personas_categoricas)
resumen_numericas(df_personas, variables_personas_numericas)
```

**Notas metodológicas**

1. Para visualizar los indicadores socioeconómicos agregados por aglomerado, fue necesario unificar las geometrías correspondientes a cada aglomerado mediante la función *dissolve*, ya que el *shapefile* original contiene múltiples polígonos por aglomerado. Esta operación permite obtener un único polígono representativo por unidad de análisis, facilitando el *merge* con los datos tabulares agregados y una correcta representación cartográfica.

2. Para vincular los datos tabulares con las geometrías de los aglomerados, fue necesario armonizar los identificadores. En las bases tabulares (hogares y personas de la EPH), los aglomerados están identificados mediante un código numérico (*AGLOMERADO*), mientras que en la capa geográfica correspondiente, el campo equivalente es *eph_codagl*, con valores en formato *string* de dos dígitos (por ejemplo, "02"). Para garantizar la compatibilidad, se convirtió el campo *AGLOMERADO* en cadena de caracteres con ceros a la izquierda (*zfill(2)*), y luego se realizó una unión por clave (*merge*) entre los datos tabulares agregados por aglomerado y la geometría espacial, generando un *GeoDataFrame* con atributos e información geográfica combinada.

3. La correspondencia entre los códigos de aglomerado en EPH y en la capa espacial fue directa, una vez armonizado el formato (*zfill*), sin necesidad de emparejamientos manuales o reinterpretaciones de nombres.

```python
# 1. Crear base de hogares únicos desde df_hogares

hogares_unicos_2 = df_hogares.drop_duplicates(subset='CODUSU')

# *** Convertir AGLOMERADO a string con 2 dígitos para evitar problemas de tipos ***

hogares_unicos_2['AGLOMERADO'] = hogares_unicos_2['AGLOMERADO'].astype(int).astype(str).str.zfill(2)

# 2. Verificar cantidad única de hogares

print(f"Cantidad de hogares únicos: {hogares_unicos_2['CODUSU'].nunique()}")
```

```python
# 1. Agregación por aglomerado a partir de hogares únicos

df_agregado = hogares_unicos_2.groupby('AGLOMERADO').agg(
    total_hogares=('ITF', 'count'),
    hogares_ITF_0=('ITF', lambda x: (x == 0).sum())
).reset_index()

# 2. Porcentaje de hogares con ITF = 0

df_agregado['porcentaje_ITF_0'] = (df_agregado['hogares_ITF_0'] / df_agregado['total_hogares']) * 100

# 3. Normalización de códigos para merge

df_agregado['AGLOMERADO'] = df_agregado['AGLOMERADO'].astype(str).str.zfill(2)  # ya debería estar ok, pero no hace mal
aglomerados['eph_codagl'] = aglomerados['eph_codagl'].astype(str).str.zfill(2)

# 4. Agrupación espacial por aglomerado (dissolve) para obtener un polígono por unidad

gdf_aglomerados = aglomerados.dissolve(by='eph_codagl', as_index=False)

# 5. Unión espacial entre datos agregados y geometrías

gdf_aglomerados = gdf_aglomerados.merge(
    df_agregado,
    left_on='eph_codagl',
    right_on='AGLOMERADO',
    how='left'
)
```

```python
from shapely.geometry import box

# 1. Definir bounding box de Argentina continental (EPSG:4326)
bbox = box(-73, -56, -53, -21)  # lon_min, lat_min, lon_max, lat_max

# 2. Crear GeoDataFrame máscara
mask = gpd.GeoDataFrame({'geometry': [bbox]}, crs='EPSG:4326')

# 3. Renombrar columnas para etiquetas consistentes
provincias = provincias.rename(columns={'NAM': 'nombre_provincia'})
departamentos = departamentos.rename(columns={'NAM': 'nombre_departamento'})
municipios = municipios.rename(columns={'NAM': 'nombre_municipio'})

# 4. Filtrar geometrías válidas (no nulas y topológicamente válidas)
provincias = provincias[provincias.geometry.notnull() & provincias.is_valid]
departamentos = departamentos[departamentos.geometry.notnull() & departamentos.is_valid]
municipios = municipios[municipios.geometry.notnull() & municipios.is_valid]
gdf_aglomerados = gdf_aglomerados[gdf_aglomerados.geometry.notnull() & gdf_aglomerados.is_valid]

# 5. Asegurar que todas las capas estén en EPSG:4326 para el clip
provincias = provincias.to_crs('EPSG:4326')
departamentos = departamentos.to_crs('EPSG:4326')
municipios = municipios.to_crs('EPSG:4326')
gdf_aglomerados = gdf_aglomerados.to_crs('EPSG:4326')

# 6. Recortar cada capa con la máscara continental
provincias = gpd.clip(provincias, mask)
departamentos = gpd.clip(departamentos, mask)
municipios = gpd.clip(municipios, mask)
gdf_aglomerados = gpd.clip(gdf_aglomerados, mask)

# 7. Reproyectar todas las capas al sistema Web Mercator EPSG:3857
provincias_3857 = provincias.to_crs(epsg=3857)
departamentos_3857 = departamentos.to_crs(epsg=3857)
municipios_3857 = municipios.to_crs(epsg=3857)
gdf_webmercator = gdf_aglomerados.to_crs(epsg=3857)

# 8. Crear figura y ejes
fig, ax = plt.subplots(figsize=(14, 12))

# 9. Plot principal: mapa coroplético del porcentaje ITF=0
gdf_webmercator.plot(
    column='porcentaje_ITF_0',
    cmap='OrRd',
    edgecolor='black',
    linewidth=0.5,
    legend=True,
    legend_kwds={'label': '% de hogares con ITF = 0'},
    ax=ax
)

# 10. Superponer límites administrativos con transparencia y distintos colores
provincias_3857.boundary.plot(ax=ax, edgecolor='gray', linewidth=0.8, alpha=0.7)
departamentos_3857.boundary.plot(ax=ax, edgecolor='lightgray', linewidth=0.5, alpha=0.5)
municipios_3857.boundary.plot(ax=ax, edgecolor='lightblue', linewidth=0.3, alpha=0.4)

# 11. Añadir etiquetas de provincias en sus centroides
for x, y, label in zip(provincias_3857.geometry.centroid.x,
                       provincias_3857.geometry.centroid.y,
                       provincias_3857['nombre_provincia']):
    ax.text(x, y, label, fontsize=8, ha='center', color='dimgray', alpha=0.7)

# 12. Añadir etiquetas con porcentaje ITF=0 para cada aglomerado
for x, y, pct in zip(gdf_webmercator.geometry.centroid.x,
                    gdf_webmercator.geometry.centroid.y,
                    gdf_webmercator['porcentaje_ITF_0']):
    ax.text(x, y, f"{pct:.1f}%", fontsize=7, ha='center', color='darkred', alpha=0.9)

# 13. Agregar mapa base
ctx.add_basemap(ax, source=ctx.providers.CartoDB.Positron)

# 14. Ajustes finales de visualización
ax.set_title('% de hogares con ITF = 0 por aglomerado urbano (EPH 2024)', fontsize=16)
ax.axis('off')

plt.tight_layout()
plt.show()
```

```python
# 1. Diccionario de regiones y sus códigos de aglomerado EPH
regiones = {
    'AMBA': ['32', '33'],
    'Región Pampeana': ['02', '03', '04', '05', '13', '34', '36', '38'],
    'Región Cuyo': ['10', '26', '27'],
    'Región Noroeste': ['18', '19', '22', '23', '25', '29'],
    'Región Noreste': ['07', '08', '12', '15'],
    'Región Patagónica': ['09', '17', '20', '30', '31', '91', '93']
}

# 2. Función para graficar una región con zoom, etiquetas y límites administrativos
def plot_region(nombre_region, codigos_region):
    subset = gdf_webmercator[gdf_webmercator['eph_codagl'].isin(codigos_region)]
    
    fig, ax = plt.subplots(figsize=(10, 10))
    
    vmin = 0
    vmax = 50
    
    # Coroplético principal
    subset.plot(
        column='porcentaje_ITF_0',
        cmap='OrRd',
        edgecolor='black',
        linewidth=0.5,
        legend=True,
        legend_kwds={'label': '% de hogares con ITF = 0'},
        vmin=vmin,
        vmax=vmax,
        ax=ax
    )
    
    # Límites administrativos intersectantes
    provincias_3857[provincias_3857.geometry.intersects(subset.unary_union)].boundary.plot(ax=ax, edgecolor='gray', linewidth=0.8, alpha=0.7)
    departamentos_3857[departamentos_3857.geometry.intersects(subset.unary_union)].boundary.plot(ax=ax, edgecolor='lightgray', linewidth=0.5, alpha=0.5)
    municipios_3857[municipios_3857.geometry.intersects(subset.unary_union)].boundary.plot(ax=ax, edgecolor='lightblue', linewidth=0.3, alpha=0.4)
    
    # Etiquetas de provincias dentro del subset
    provincias_sub = provincias_3857[provincias_3857.geometry.intersects(subset.unary_union)]
    for x, y, label in zip(provincias_sub.geometry.centroid.x,
                           provincias_sub.geometry.centroid.y,
                           provincias_sub['nombre_provincia']):
        ax.text(x, y, label, fontsize=8, ha='center', color='dimgray', alpha=0.7)
    
    # Etiquetas de porcentaje ITF=0 por aglomerado
    for x, y, pct in zip(subset.geometry.centroid.x,
                         subset.geometry.centroid.y,
                         subset['porcentaje_ITF_0']):
        ax.text(x, y, f"{pct:.1f}%", fontsize=7, ha='center', color='darkred', alpha=0.9)
    
    # Mapa base y zoom
    ctx.add_basemap(ax, source=ctx.providers.CartoDB.Positron)
    
    xmin, ymin, xmax, ymax = subset.total_bounds
    ax.set_xlim(xmin, xmax)
    ax.set_ylim(ymin, ymax)
    
    ax.set_title(f'{nombre_region}: % de hogares con ITF = 0', fontsize=16)
    ax.axis('off')
    
    plt.tight_layout()
    plt.show()

# 3. Graficar todas las regiones
for nombre, codigos in regiones.items():
    plot_region(nombre, codigos)
```

Los mapas presentados permiten observar la distribución geográfica del porcentaje de hogares con ITF = 0 en los aglomerados urbanos de la EPH 2024.

## 4. Análisis Espacial Exploratorio

Para profundizar el análisis cartográfico, se aplicó un conjunto de estadísticos espaciales. En primer lugar, se calculó el Índice Global de Moran, que permite evaluar si el porcentaje de hogares sin ingresos registrados (ITF = 0) presenta una distribución agregada, dispersa o aleatoria en el espacio. Un índice significativamente positivo indicaría patrones de agrupamiento espacial.

A nivel local, se utilizó el Índice Local de Moran (LISA) para identificar aglomerados espacialmente atípicos:

- **High-High (Q1)**: zonas con valores elevados cercanas a otras zonas también elevadas (*hotspots*).
- **Low-Low (Q3)**: zonas con valores bajos rodeadas de otras bajas (*coldspots*).
- **High-Low (Q4)** y **Low-High (Q2)**: zonas que difieren de sus vecinas, indicando posibles *outliers* espaciales.

Dado que los aglomerados urbanos relevados por la EPH no presentan contigüidad espacial directa (no comparten bordes), se utilizó una matriz de vecinos basada en los cinco aglomerados más cercanos (*K*-vecinos). Esta metodología asegura que cada unidad espacial tenga vínculos definidos, evitando la generación de islas y permitiendo una evaluación robusta de la autocorrelación espacial.

Para contextualizar espacialmente los resultados, se superpuso la capa ReNaBaP de barrios populares, a fin de evaluar si existe una coincidencia territorial entre las áreas de mayor concentración de hogares sin ingresos y zonas de vulnerabilidad habitacional.

Adicionalmente, se calculó la proporción de aglomerados clasificados como High-High (hotspots) que presentan intersección con al menos un barrio popular del ReNaBaP, con el objetivo de aportar evidencia cuantitativa de dicha coincidencia espacial.

```python
import libpysal
import esda
import splot.esda as sp_esda
from libpysal.weights import KNN
import contextily as ctx
import matplotlib.pyplot as plt

# 1. Asegurar CRS planar
gdf = gdf_webmercator.copy()
renabap_web_3857 = renabap.to_crs(epsg=3857)

# 2. Recorte si no se hizo
renabap_web_3857 = gpd.clip(
    renabap_web_3857,
    gpd.GeoDataFrame(geometry=[bbox], crs='EPSG:4326').to_crs(epsg=3857)
)

# 3. Eliminar valores nulos
gdf = gdf.dropna(subset=['porcentaje_ITF_0']).reset_index(drop=True)

# 4. Matriz de vecinos tipo KNN (5 vecinos más cercanos)
w = KNN.from_dataframe(gdf, k=5)
w.transform = 'r'

# 5. Índice de Moran Global
moran = esda.Moran(gdf['porcentaje_ITF_0'], w)
print(f"Moran I global: {moran.I:.4f}, p‑valor: {moran.p_norm:.4f}")

# 6. Índice LISA
lisa = esda.Moran_Local(gdf['porcentaje_ITF_0'], w)
gdf['lisa_I'] = lisa.Is
gdf['lisa_p'] = lisa.p_sim
gdf['lisa_q'] = lisa.q
gdf['significativo'] = gdf['lisa_p'] < 0.05

# 6.1. Describir la matriz KNN (estructura espacial)
# Tamaño promedio de vecinos por aglomerado
vecinos_por_obs = w.cardinalities.values()
print(f"Promedio de vecinos por aglomerado: {sum(vecinos_por_obs)/len(vecinos_por_obs):.2f}")

# 6.2. Contar clústeres LISA significativos por tipo (Q1–Q4)
# Conteo de clústeres significativos por cuadrante
conteo_clusters = gdf[gdf['significativo']]['lisa_q'].value_counts().sort_index()
for q in [1, 2, 3, 4]:
    cant = conteo_clusters.get(q, 0)
    print(f"Q{q} – {'High-High' if q==1 else 'Low-High' if q==2 else 'Low-Low' if q==3 else 'High-Low'}: {cant} aglomerados")

# 6.3. Porcentaje de aglomerados en clústeres significativos
total_aglomerados = len(gdf)
total_significativos = gdf['significativo'].sum()
print(f"{total_significativos} de {total_aglomerados} aglomerados ({100*total_significativos/total_aglomerados:.1f}%) tienen clústeres LISA significativos.")

# 6.4. Porcentaje de clústeres High-High (Q1) que intersectan ReNaBaP
hh_clusters = gdf[(gdf['lisa_q'] == 1) & (gdf['significativo'])]
hh_with_renabap = hh_clusters[hh_clusters.intersects(renabap_web_3857.unary_union)]
pct_hh_intersecta = 100 * len(hh_with_renabap) / len(hh_clusters) if len(hh_clusters) > 0 else 0
print(f"{pct_hh_intersecta:.1f}% de los clústeres HH intersectan al menos un barrio ReNaBaP.")

# 7. Mapa de clústeres significativos
fig, ax = plt.subplots(figsize=(10, 10))
colores = {1: 'darkred', 2: 'blue', 3: 'darkblue', 4: 'orange'}
clusters = gdf[gdf['significativo']]

for q, color in colores.items():
    clusters[clusters['lisa_q'] == q].plot(ax=ax, color=color, edgecolor='black', linewidth=0.5, label=f'Q{q}')

# Agregar límites y ReNaBaP
provincias_3857.boundary.plot(ax=ax, edgecolor='gray', linewidth=0.5, alpha=0.5)
renabap_web_3857.plot(ax=ax, color='none', edgecolor='purple', linewidth=1, alpha=0.7, label='ReNaBaP')

# Mapa base
ctx.add_basemap(ax, source=ctx.providers.CartoDB.Positron)
ax.set_title('Hotspots y clústeres LISA de hogares con ITF = 0', fontsize=14)
ax.axis('off')
ax.legend(title='Cuadrantes LISA', loc='lower left')
plt.tight_layout()
plt.show()
```

**Análisis de autocorrelación espacial (Global Moran's I y LISA)**

Se utilizó el índice de autocorrelación espacial de Moran (*Global Moran's I*) para evaluar si la distribución espacial del porcentaje de hogares con ITF igual a cero presenta patrones de agrupamiento geográfico significativos en los aglomerados urbanos. El resultado obtenido fue:

- Índice de Moran I global: 0.3401

- p‑valor: 0.0000

Este valor positivo y estadísticamente significativo indica que los valores similares tienden a agruparse espacialmente: es decir, existen conglomerados de aglomerados urbanos con altos (o bajos) porcentajes de hogares sin ingresos declarados que no se distribuyen al azar, sino que forman patrones geográficos consistentes. Un valor de Moran cercano a 1 implicaría una autocorrelación espacial muy fuerte; el valor obtenido (0.34) sugiere una autocorrelación moderada pero robusta.

Para identificar la localización de estos conglomerados, se aplicó el análisis local de autocorrelación espacial (LISA), que permite distinguir entre distintos tipos de agrupamientos. El análisis LISA clasifica los aglomerados en cuatro cuadrantes del plano de Moran:

- Q1 – High-High (HH): aglomerados con valores altos rodeados por vecinos también con valores altos.

- Q2 – Low-High (LH): valores bajos rodeados por vecinos con valores altos.

- Q3 – Low-Low (LL): valores bajos rodeados por vecinos también bajos.

- Q4 – High-Low (HL): valores altos rodeados por vecinos con valores bajos.

El análisis identificó los siguientes clústeres LISA significativos (p < 0.05):

- Q1 – High-High: 8 aglomerados

- Q3 – Low-Low: 6 aglomerados

- Q2 y Q4: 0 aglomerados

Esto implica que el 43.8% de los aglomerados (14 de 32) presentan una estructura espacial significativamente autocorrelacionada, ya sea como conglomerados de alta intensidad del fenómeno (HH) o de baja intensidad (LL). La ausencia de aglomerados en los cuadrantes LH y HL sugiere que no se observan valores atípicos espaciales (es decir, aglomerados "desalineados" respecto de sus vecinos inmediatos).

El número promedio de vecinos considerados por aglomerado fue de 5.00, en base a una matriz de contigüidad de primer orden (reina), lo cual garantiza una conectividad razonable para captar interacciones espaciales sin sobredimensionar la vecindad.

Una observación relevante es que el 100% de los clústeres High-High intersectan al menos un barrio ReNaBaP, lo cual sugiere una posible relación entre la concentración de hogares sin ingresos y la localización de asentamientos populares reconocidos por el Estado. Esta superposición refuerza la hipótesis de que el no registro de ingresos en la EPH puede estar asociado a formas de informalidad estructural y exclusión residencial.

## 5. Interpretación y discusión

Los resultados del análisis espacial confirman que la omisión de ingresos en la EPH no se distribuye de forma aleatoria, sino que se concentra en determinados aglomerados urbanos con patrones espaciales identificables. El índice de Moran Global evidencia una autocorrelación positiva moderada, mientras que el análisis LISA permite identificar clústeres significativos tanto de alta como de baja intensidad del fenómeno. La coincidencia absoluta entre los clústeres de alta intensidad (High-High) y la presencia de barrios populares (ReNaBaP) refuerza la hipótesis de que existe una asociación estructural entre la informalidad habitacional y la omisión de ingresos en la encuesta.

Esta asociación sugiere que en contextos de alta precariedad urbana (donde predominan el empleo informal, la desconfianza institucional, la debilidad censal o la marginalidad residencial) puede existir un subregistro y no registro sistemático de ingresos. No necesariamente porque no existan ingresos, sino porque los hogares no los declaran, no los identifican como tales, o no son captados adecuadamente por la encuesta.

Por otro lado, la ausencia de *outliers* espaciales (cuadrantes LH o HL) puede indicar que la dinámica territorial del fenómeno está más relacionada con entornos estructuralmente homogéneos (es decir, los aglomerados se comportan como sistemas coherentes más que como unidades aisladas).

Estos hallazgos ponen en cuestión la supuesta representatividad territorial homogénea de la EPH. Aunque diseñada para capturar realidades urbanas diversas, su arquitectura muestral parece enfrentar límites metodológicos en contextos de alta vulnerabilidad o informalidad. En estos casos, los indicadores construidos a partir de los datos de ingreso deben ser leídos con cautela, especialmente cuando se utilizan para estimar pobreza o evaluar políticas públicas.

En este sentido, el trabajo aporta una lectura territorial del no registro estadístico de ingresos, y se inscribe en un conjunto más amplio de estudios que interpelan los supuestos de neutralidad técnica en la producción de información oficial. Lejos de ser meramente un "error muestral", el no registro de ingresos revela las tensiones entre las herramientas estadísticas, la informalidad estructural y la exclusión social en contextos urbanos fragmentados.
